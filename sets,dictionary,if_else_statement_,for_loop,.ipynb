**1.Explain with an example each when to use a for loop and a while loop?**

**for loop**-you know exactly how many times the loop needs to run before it even begins.

 **Ex-** cart_items = [10.99, 5.50, 20.00]

   total_cost = 0

for price in cart_items:
    total_cost += price

print(f"Total: ${total_cost}")  

**While loop**- you do not know how many times the code will need to repeat.

**Ex**- countdown=int(input("enter the number"))
while countdown > 0:
    print(countdown)
    countdown = countdown - 1  # Reduces the count by 1 each time

print("Blast off!")
